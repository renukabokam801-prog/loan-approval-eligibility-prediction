## 5. Statistical Testing

### Objective
To identify whether applicant characteristics have a statistically significant relationship with historical loan approval decisions.

### Statistical Methods

### 1. Chi-Square Test: Credit History vs Loan Status

We use the Chi-Square Test of Independence to determine whether Credit History is significantly associated with Loan Status.

```python
import pandas as pd
from scipy.stats import chi2_contingency

df = pd.read_csv("../data/raw/train_u6lujuX_CVtuZ9i.csv")

contingency_table = pd.crosstab(df["Credit_History"], df["Loan_Status"])

contingency_table
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

### Result and Interpretation

- Chi-Square Statistic: 174.6373
- P-value: 7.1848e-40
- Degrees of Freedom: 1
- Significance Level: 0.05

**Decision:** Since the p-value is less than 0.05, we reject the null hypothesis.

**Conclusion:** There is a statistically significant association between Credit History and Loan Status. Applicants with a positive Credit History had a considerably higher loan approval rate in this dataset.

This result indicates that Credit History may be an important feature for loan approval prediction. However, statistical association does not imply causation.

### 8. Mann–Whitney U Test: Applicant Income vs Loan Status

**Objective:** To determine whether ApplicantIncome differs significantly between approved and rejected applicants.

**Hypotheses:**
- H0: ApplicantIncome distributions are the same for approved and rejected applicants.
- H1: ApplicantIncome distributions are different.

**Significance Level:** 0.05

**Statistical Test:** Mann–Whitney U Test

```python
from scipy.stats import mannwhitneyu

approved = df[df["Loan_Status"] == "Y"]["ApplicantIncome"]
rejected = df[df["Loan_Status"] == "N"]["ApplicantIncome"]

statistic, p_value = mannwhitneyu(approved, rejected)

print("Mann-Whitney U Statistic:", statistic)
print("P-value:", p_value)
```

### Result and Interpretation

- Mann-Whitney U Statistic: 40302.0
- P-value: 0.9181

**Decision:**

Since the p-value (0.9181) is greater than 0.05, we fail to reject the null hypothesis (H0).

**Conclusion:**

There is no statistically significant difference in ApplicantIncome distributions between approved and rejected applicants in this dataset.

Therefore, ApplicantIncome alone does not show a significant difference between the two groups.

**Note:** This does not mean ApplicantIncome is useless for prediction. It may be useful when combined with other features.

## 15. Multiple Testing Correction

### Objective
To reduce the chance of false-positive results when performing multiple statistical tests.

### Method
Bonferroni Correction

### Significance Level
0.05

The corrected p-value is calculated by multiplying each original p-value by the total number of tests.

A corrected p-value less than or equal to 0.05 is considered statistically significant.

```python
results = pd.DataFrame({
    "Feature": [
        "Credit_History", "Education", "Married", "Property_Area",
        "Gender", "Self_Employed", "Dependents",
        "ApplicantIncome", "CoapplicantIncome", "LoanAmount",
        "Loan_Amount_Term", "TotalIncome", "LoanToIncomeRatio"
    ],
    "P_Value": [
        7.184759548750746e-40,
        0.04309962129357355,
        0.03439381301579988,
        0.0021360187811644937,
        0.7086529816451106,
        1.0,
        0.3678506740863211,
        0.91811428057985,
        0.38687350990802183,
        0.39520911423457217,
        0.5181103836516355,
        0.7128351432717865,
        0.5764420723804118
    ]
})

results["Adjusted_P_Value"] = (
    results["P_Value"] * len(results)
).clip(upper=1)

results["Significant"] = results["Adjusted_P_Value"] <= 0.05

results
```

## 16. Interpretation of Multiple Testing Correction

After applying the Bonferroni correction:

- Credit_History remains statistically significant (adjusted p-value = 9.34e-39).
- Property_Area remains statistically significant (adjusted p-value = 0.0278).
- Education and Married are no longer statistically significant.
- All other tested features remain statistically insignificant.

### Conclusion

Credit_History shows the strongest statistical association with historical loan approval decisions.

Property_Area also shows a statistically significant association.

These findings help us understand the dataset, but they do not independently determine which features will perform best in a machine learning model.
