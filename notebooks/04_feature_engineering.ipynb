#  Loan Approval Eligibility Prediction

## 4. Feature Engineering

### Objective
To create new meaningful features from existing dataset columns that may help improve the performance of the loan approval prediction model.

### Features to be Created
1. TotalIncome
2. LoanToIncomeRatio

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/train_u6lujuX_CVtuZ9i.csv")

df.head()
```

```python
df.columns
```

```python
df["TotalIncome"] = df["ApplicantIncome"] + df["CoapplicantIncome"]

df[["ApplicantIncome", "CoapplicantIncome", "TotalIncome"]].head()
```

### 1. Total Income

TotalIncome is calculated by adding ApplicantIncome and CoapplicantIncome.

It represents the combined income of the applicant and co-applicant.

```python
df["LoanToIncomeRatio"] = df["LoanAmount"] / df["TotalIncome"]

df[["LoanAmount", "TotalIncome", "LoanToIncomeRatio"]].head()
```

### 2. Loan-to-Income Ratio

LoanToIncomeRatio is calculated by dividing LoanAmount by TotalIncome.

It helps us understand the requested loan amount in relation to the applicant's combined income.

A higher ratio indicates a larger loan amount relative to the applicant's income.

```python
df[["TotalIncome", "LoanToIncomeRatio"]].describe()
```

### Observations

1. TotalIncome was successfully calculated for all 614 applicants.
2. LoanToIncomeRatio is available for 592 applicants because some LoanAmount values are missing.
3. These newly created features may help the model understand the relationship between income and loan amount.

```python
df["Dependents"].value_counts(dropna=False)
```

### 3. Dependents Feature

The Dependents column contains four categories: 0, 1, 2, and 3+.

Since 3+ represents three or more dependents, we retain this feature as categorical rather than converting it into a numerical value.

Missing values will be handled during model preprocessing.

```python
df[["ApplicantIncome", "CoapplicantIncome", "TotalIncome",
    "LoanAmount", "LoanToIncomeRatio", "Dependents"]].head()
```

## 5. Conclusion

Feature Engineering was performed to create meaningful features from the existing dataset.

The following features were created:

1. TotalIncome – Combined income of applicant and co-applicant.
2. LoanToIncomeRatio – Loan amount relative to total income.

The Dependents feature was retained as a categorical variable.

These features will be evaluated during model building to determine their usefulness in predicting loan approval eligibility.
