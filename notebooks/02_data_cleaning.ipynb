# Loan Approval Eligibility Prediction

## 2. Data Cleaning

### Objective
To identify and handle missing values and prepare the dataset for further analysis.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/train_u6lujuX_CVtuZ9i.csv")

df.head()
```

### 1. Identifying Missing Values
We check missing values in each column to identify incomplete records.

```python
df.isnull().sum()
```

### 2. Handling Missing Values
Missing values can affect data analysis and model performance.
We will use suitable imputation methods to fill them.

```python
df.isnull().sum().sum()
```

### 3. Identifying Column Types
We separate numerical and categorical columns to apply suitable imputation methods.

```python
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(include="object").columns

print("Numerical columns:", list(num_cols))
print("Categorical columns:", list(cat_cols))
```

### 4. Selecting Columns for Imputation
Numerical missing values will be filled using the median, while categorical missing values will be filled using the mode.

```python
num_cols = df.select_dtypes(include="number").columns.drop("Credit_History")
cat_cols = df.select_dtypes(include="str").columns.drop(["Loan_ID", "Loan_Status"])

print("Numerical:", list(num_cols))
print("Categorical:", list(cat_cols))
```

### 5. Numerical Imputation
We use the median to fill missing numerical values because it is less affected by outliers.

```python
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

### 6. Categorical Imputation
We use the mode (most frequent value) to fill missing categorical values.

```python
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

### 7. Credit History Imputation
Missing credit history values are filled using the mode because it is a binary categorical feature (0 or 1).

```python
df["Credit_History"] = df["Credit_History"].fillna(df["Credit_History"].mode()[0])
```

### 8. Verifying Missing Values
We check whether all missing values have been successfully handled.

```python
df.isnull().sum()
```

### 9. Cleaning Summary
All missing values have been handled using median and mode imputation.
The dataset is now free from missing values.

```python
df.shape
```

We display the cleaned dataset to verify the changes.

We save the cleaned dataset for use in further analysis.
